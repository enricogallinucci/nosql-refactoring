```python
import pandas as pd
```

```python
filename = f"../results/results_2003.csv"
log_df = pd.read_csv(filename)
```

We exclude the first 1k executions to try and fill all the cache, to then have a fully comparable set of queries

```python
log_df = log_df.iloc[100:]
```

```python
log_df.shape
```

```python
log_df.head()
```

```python
query_count_summary = log_df['QueryType'].value_counts().reset_index()
query_count_summary.columns = ['QueryType', 'Count']
total_count = query_count_summary['Count'].sum()
query_count_summary['Percentage'] = (query_count_summary['Count'] / total_count) * 100

query_count_summary
```

```python
execution_time_summary = log_df.groupby('QueryType')['Execution Time'].sum().reset_index()
total_execution_time = execution_time_summary['Execution Time'].sum()
execution_time_summary['Percentage'] = (execution_time_summary['Execution Time'] / total_execution_time) * 100


execution_time_summary
```

```python
summary = log_df.groupby('QueryType').agg(
    Total_Execution_Time=('Execution Time', 'sum'),
    Execution_Count=('QueryType', 'count'),
    Mean_Execution_Time=('Execution Time', 'mean'),
    Std_Dev_Execution_Time=('Execution Time', 'std')
)

summary['Total_Execution_Percentage'] = (summary['Total_Execution_Time'] / summary['Total_Execution_Time'].sum()) * 100
# summary['Mean_Execution_Time_Percentage'] = (summary['Mean_Execution_Time'] / summary['Mean_Execution_Time'].sum()) * 100

summary
```
