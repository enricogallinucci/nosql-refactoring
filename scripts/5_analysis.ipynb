```python
# Script to get list of tables involved in queries in the yml files

import re
import yaml

# Load the YAML file
with open("../queries/2023-optimized.yml", 'r') as file:
    data = yaml.safe_load(file)

# Extract SQL queries and identify table names using regex
table_names = set()
for entry in data:
    if 'query' in entry:
        query = entry['query'].lower()
        # Match potential table names in FROM or JOIN clauses
        tables = re.findall(r'\bFROM\s+([\w.]+)|\bJOIN\s+([\w.]+)', query, re.IGNORECASE)
        # Flatten and add non-empty matches to the set
        table_names.update(filter(None, [match for pair in tables for match in pair]))

for t in sorted(table_names):
    print(t)
```
